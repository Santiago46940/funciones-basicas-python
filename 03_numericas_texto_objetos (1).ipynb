### `abs()`
**Descripción:** Valor absoluto.

**Sintaxis:** `abs(x)`

```python
print(abs(-7), abs(3.2), abs(-2+0j))
```

### `round()`
**Descripción:** Redondea a n decimales (redondeo bancario en empates).

**Sintaxis:** `round(x, ndigitos=None)`

```python
print(round(3.14159, 2), round(2.5), round(3.5), round(1234, -2))
```

### `sum()`
**Descripción:** Suma los elementos de un iterable.

**Sintaxis:** `sum(iterable, start=0)`

```python
print(sum([1, 2, 3]), sum([0.1, 0.2], 10))
```

### `min()`
**Descripción:** Menor elemento (admite `key`).

**Sintaxis:** `min(iterable, key=None) | min(a, b, ...)`

```python
print(min(4, 2, 9), min(["pera", "fe"], key=len))
```

### `max()`
**Descripción:** Mayor elemento (admite `key`).

**Sintaxis:** `max(iterable, key=None) | max(a, b, ...)`

```python
print(max([4, 2, 9]), max({"a": 3, "b": 7}, key=lambda k: {"a": 3, "b": 7}[k]))
```

### `pow()`
**Descripción:** Potencia; con tercer argumento calcula módulo.

**Sintaxis:** `pow(base, exp, mod=None)`

```python
print(pow(2, 10), pow(2, 10, 1000))
```

### `divmod()`
**Descripción:** Devuelve cociente y residuo a la vez.

**Sintaxis:** `divmod(a, b)`

```python
minutos, segundos = divmod(367, 60)
print(minutos, segundos)
```

### `chr()`
**Descripción:** Carácter correspondiente a un código Unicode.

**Sintaxis:** `chr(entero)`

```python
print(chr(65), chr(97), chr(9731))
```

### `ord()`
**Descripción:** Código Unicode de un carácter (inversa de `chr`).

**Sintaxis:** `ord(caracter)`

```python
print(ord("A"), ord("ñ"))
# Cifrado César simple
print("".join(chr((ord(c) - 97 + 3) % 26 + 97) for c in "hola"))
```

### `bin()`
**Descripción:** Representación binaria de un entero.

**Sintaxis:** `bin(entero)`

```python
print(bin(10), bin(255))
```

### `hex()`
**Descripción:** Representación hexadecimal de un entero.

**Sintaxis:** `hex(entero)`

```python
print(hex(255), hex(4096))
```

### `dir()`
**Descripción:** Lista los atributos y métodos de un objeto.

**Sintaxis:** `dir(objeto)`

```python
print([m for m in dir(str) if not m.startswith("_")][:10])
```

### `hasattr / getattr()`
**Descripción:** Verifica y obtiene un atributo por su nombre (texto).

**Sintaxis:** `hasattr(obj, 'attr') ; getattr(obj, 'attr', defecto)`

```python
class Persona:
    nombre = "Ana"
p = Persona()
print(hasattr(p, "nombre"), hasattr(p, "edad"))
print(getattr(p, "edad", "desconocida"))
```

### `callable()`
**Descripción:** Indica si un objeto se puede invocar como función.

**Sintaxis:** `callable(objeto)`

```python
print(callable(len), callable("texto"), callable(lambda: 0))
```

### `open()`
**Descripción:** Abre un archivo; se usa con `with` para cerrarlo automáticamente.

**Sintaxis:** `open(ruta, modo='r', encoding=None)`

```python
with open("demo.txt", "w", encoding="utf-8") as f:
    f.write("línea 1\nlínea 2\n")

with open("demo.txt", encoding="utf-8") as f:
    for linea in f:
        print(linea.strip())

import os; os.remove("demo.txt")
```
