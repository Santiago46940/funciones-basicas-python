### `print()`
**Descripción:** Muestra valores en pantalla. Admite `sep` y `end` para controlar el formato.

**Sintaxis:** `print(*objetos, sep=' ', end='\n')`

```python
print("Hola", "mundo")
print("a", "b", "c", sep="-")
print("sin salto", end=" | ")
print("continúa")
```

### `input()`
**Descripción:** Lee texto escrito por el usuario (siempre devuelve `str`).

**Sintaxis:** `input(mensaje)`

```python
# nombre = input("¿Cómo te llamas? ")   # en un script interactivo
nombre = "Ana"  # simulamos la respuesta para que el notebook se ejecute solo
edad = int("20")  # input() devuelve texto: hay que convertirlo
print(f"Hola {nombre}, el próximo año tendrás {edad + 1}")
```

### `type()`
**Descripción:** Devuelve el tipo de un objeto.

**Sintaxis:** `type(objeto)`

```python
for v in [10, 3.14, "hola", True, [1, 2], (1, 2), {"a": 1}, {1, 2}, None]:
    print(f"{v!r:12} -> {type(v).__name__}")
```

### `isinstance()`
**Descripción:** Comprueba si un objeto es de un tipo (o de varios).

**Sintaxis:** `isinstance(objeto, tipo | (tipos))`

```python
print(isinstance(5, int))
print(isinstance(5, (str, float)))
print(isinstance(True, int))  # bool hereda de int
```

### `id()`
**Descripción:** Devuelve el identificador único (dirección en CPython) de un objeto.

**Sintaxis:** `id(objeto)`

```python
a = [1, 2]
b = a          # misma referencia
c = a.copy()   # copia
print(id(a) == id(b), id(a) == id(c))
```

### `int()`
**Descripción:** Convierte a entero (también admite base).

**Sintaxis:** `int(x, base=10)`

```python
print(int("42"), int(7.9), int("1010", 2), int("ff", 16))
```

### `float()`
**Descripción:** Convierte a número decimal.

**Sintaxis:** `float(x)`

```python
print(float("3.5"), float(2), float("1e3"))
```

### `str()`
**Descripción:** Convierte cualquier objeto a cadena de texto.

**Sintaxis:** `str(objeto)`

```python
print(str(123) + "4", str([1, 2]), str(None))
```

### `bool()`
**Descripción:** Convierte a booleano. Son `False`: 0, '', [], {}, None.

**Sintaxis:** `bool(x)`

```python
for v in [0, 1, "", "a", [], [0], None]:
    print(f"{v!r:5} -> {bool(v)}")
```

### `repr()`
**Descripción:** Representación 'oficial' (sin ambigüedad) de un objeto; útil para depurar.

**Sintaxis:** `repr(objeto)`

```python
s = "hola\nmundo"
print(str(s))
print(repr(s))
```

### `format()`
**Descripción:** Da formato a un valor según una especificación.

**Sintaxis:** `format(valor, especificacion)`

```python
print(format(3.14159, ".2f"))
print(format(1234567, ","))
print(format(0.256, ".1%"))
print(format(42, "08b"))
```

### `help()`
**Descripción:** Muestra la documentación de una función u objeto.

**Sintaxis:** `help(objeto)`

```python
help(len)
```
