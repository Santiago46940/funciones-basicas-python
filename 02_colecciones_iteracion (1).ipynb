Funciones cubiertas: `len`, `list`, `tuple`, `dict`, `set`, `range`, `sorted`, `reversed`, `enumerate`, `zip`, `map`, `filter`, `any`, `all`, `iter`/`next`.

### `len()`
**Descripción:** Cantidad de elementos de una colección o caracteres de un texto.

**Sintaxis:** `len(objeto)`

```python
print(len("Python"), len([1, 2, 3]), len({"a": 1, "b": 2}))
```

### `list()`
**Descripción:** Crea una lista a partir de un iterable.

**Sintaxis:** `list(iterable)`

```python
print(list("abc"), list(range(3)), list((1, 2)))
```

### `tuple()`
**Descripción:** Crea una tupla (secuencia inmutable).

**Sintaxis:** `tuple(iterable)`

```python
t = tuple([1, 2, 3])
print(t)
# t[0] = 9  # TypeError: las tuplas no se modifican
```

### `dict()`
**Descripción:** Crea un diccionario (pares clave-valor).

**Sintaxis:** `dict(**kwargs) | dict(iterable_de_pares)`

```python
d = dict(nombre="Luis", edad=21)
print(d, d["nombre"], d.get("correo", "sin correo"))
print(dict(zip(["x", "y"], [1, 2])))
```

### `set()`
**Descripción:** Crea un conjunto (sin duplicados, sin orden).

**Sintaxis:** `set(iterable)`

```python
s = set([1, 2, 2, 3, 3, 3])
print(s, {1, 2} | {2, 3}, {1, 2} & {2, 3})
```

### `range()`
**Descripción:** Secuencia de enteros perezosa: inicio, fin (excluido) y paso.

**Sintaxis:** `range(inicio, fin, paso)`

```python
print(list(range(5)), list(range(2, 10, 3)), list(range(5, 0, -2)))
```

### `sorted()`
**Descripción:** Devuelve una lista nueva ordenada (no modifica el original).

**Sintaxis:** `sorted(iterable, key=None, reverse=False)`

```python
print(sorted([3, 1, 2]))
print(sorted(["pera", "fe", "manzana"], key=len))
print(sorted([3, 1, 2], reverse=True))
```

### `reversed()`
**Descripción:** Devuelve un iterador que recorre en orden inverso.

**Sintaxis:** `reversed(secuencia)`

```python
print(list(reversed([1, 2, 3])), "".join(reversed("hola")))
```

### `enumerate()`
**Descripción:** Recorre un iterable obteniendo índice y valor.

**Sintaxis:** `enumerate(iterable, start=0)`

```python
for i, fruta in enumerate(["uva", "kiwi", "mango"], start=1):
    print(i, fruta)
```

### `zip()`
**Descripción:** Combina varios iterables elemento a elemento.

**Sintaxis:** `zip(*iterables)`

```python
nombres = ["Ana", "Luis"]
notas = [4.5, 3.8]
for n, nota in zip(nombres, notas):
    print(n, nota)
```

### `map()`
**Descripción:** Aplica una función a cada elemento de un iterable.

**Sintaxis:** `map(funcion, iterable)`

```python
print(list(map(str.upper, ["a", "b"])))
print(list(map(lambda x: x ** 2, [1, 2, 3])))
```

### `filter()`
**Descripción:** Conserva los elementos para los que la función devuelve verdadero.

**Sintaxis:** `filter(funcion, iterable)`

```python
print(list(filter(lambda x: x % 2 == 0, range(10))))
```

### `any()`
**Descripción:** `True` si al menos un elemento es verdadero.

**Sintaxis:** `any(iterable)`

```python
print(any([0, 0, 1]), any([]))
print(any(n < 0 for n in [3, -1, 5]))
```

### `all()`
**Descripción:** `True` si todos los elementos son verdaderos.

**Sintaxis:** `all(iterable)`

```python
print(all([1, 2, 3]), all([1, 0]), all([]))
print(all(n > 0 for n in [3, 1, 5]))
```

### `iter / next()`
**Descripción:** `iter` crea un iterador; `next` obtiene el siguiente elemento.

**Sintaxis:** `iter(iterable) ; next(iterador, valor_por_defecto)`

```python
it = iter([10, 20])
print(next(it), next(it), next(it, "fin"))
```
